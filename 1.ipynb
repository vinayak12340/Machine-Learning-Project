```python
# Remove duplicate rows
df = df.drop_duplicates().copy()

# Remove empty or whitespace-only reviews
df = df[
    df["verified_reviews"].astype(str).str.strip().ne("")
].copy()

# Convert reviews to lowercase
df["verified_reviews"] = (
    df["verified_reviews"].astype(str).str.lower()
)

print("Dataset shape after cleaning:", df.shape)

print("\nEmpty reviews remaining:",
      df["verified_reviews"].str.strip().eq("").sum())
```

```python
df["sentiment"] = pd.cut(
    df["rating"],
    bins=[0, 2, 3, 5],
    labels=["Negative", "Neutral", "Positive"],
    include_lowest=True
)
df
```

```python
import pandas as pd
import re
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

# Function for tokenization and lemmatization
def preprocess_text(text):
    # Convert to lowercase
    text = str(text).lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Tokenization
    tokens = word_tokenize(text)

    # Lemmatization
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # Join tokens back into text
    return " ".join(tokens)

# Apply preprocessing
df["processed_reviews"] = df["verified_reviews"].apply(preprocess_text)

# Display original and processed reviews
display(df[["verified_reviews", "processed_reviews", "sentiment"]].head(10))

```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Bag of Words
bow_vectorizer = CountVectorizer()
X_bow = bow_vectorizer.fit_transform(df["processed_reviews"])

print("Bag of Words Shape:", X_bow.shape)

# TF-IDF
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(df["processed_reviews"])

print("TF-IDF Shape:", X_tfidf.shape)

# Target variable: 3-class sentiment
y = df["sentiment"]

print("\nTarget Distribution:")
print(y.value_counts())
```

```python


from sklearn.model_selection import train_test_split

X = df["processed_reviews"]
y = df["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.40,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# 1. Convert text into Bag of Words features
bow = CountVectorizer()

X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)

# 2. Train Multinomial Naive Bayes
mnb = MultinomialNB()
mnb.fit(X_train_bow, y_train)

# 3. Predict sentiment
y_pred = mnb.predict(X_test_bow)

# 4. Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python

from sklearn.naive_bayes import MultinomialNB
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Calculate balanced weights for training samples only
sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train
)

# Train weighted MNB using the existing BoW features
mnb_balanced = MultinomialNB()
mnb_balanced.fit(
    X_train_bow,
    y_train,
    sample_weight=sample_weights
)

# Predict on the unchanged test set
y_pred_mnb_balanced = mnb_balanced.predict(X_test_bow)

# Evaluate
print("MNB + BoW (Balanced Sample Weights)")
print("Accuracy:", round(
    accuracy_score(y_test, y_pred_mnb_balanced), 4
))
print("Macro F1:", round(
    f1_score(y_test, y_pred_mnb_balanced, average="macro"), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_mnb_balanced,
    labels=["Negative", "Neutral", "Positive"],
    zero_division=0
))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# 1. Initialize Logistic Regression with class balancing
lr_bow = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

# 2. Train model
lr_bow.fit(X_train_bow, y_train)

# 3. Predict on test data
y_pred_lr_bow = lr_bow.predict(X_test_bow)

# 4. Evaluate model
print("Logistic Regression + Bag of Words")

print("Accuracy:", round(
    accuracy_score(y_test, y_pred_lr_bow), 4
))

print("Macro Precision:", round(
    precision_score(y_test, y_pred_lr_bow, average="macro", zero_division=0), 4
))

print("Macro Recall:", round(
    recall_score(y_test, y_pred_lr_bow, average="macro", zero_division=0), 4
))

print("Macro F1-score:", round(
    f1_score(y_test, y_pred_lr_bow, average="macro", zero_division=0), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_lr_bow,
    labels=["Negative", "Neutral", "Positive"],
    zero_division=0
))
```
